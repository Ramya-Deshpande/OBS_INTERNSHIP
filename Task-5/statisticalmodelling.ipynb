```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("musicdata.csv")
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df=df.drop(df.columns[0],axis=1)
df
```

```python
categorical_columns=df.select_dtypes('object').columns
categorical_columns
```

```python
df[categorical_columns]=df[categorical_columns].fillna('Unknown')
categorical_columns
```

```python
numerical_columns=df.select_dtypes(include=['int64','float64']).columns
numerical_columns
```

```python
for col in numerical_columns:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col],bins=30,kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
df[numerical_columns].skew()
```

```python
fill_values = {}

for col in numerical_columns:
    skewness = df[col].skew()
    
    if abs(skewness) > 0.5:  
        fill_values[col] = df[col].median()
    elif df[col].nunique() < 10: 
        fill_values[col] = df[col].mode()[0]
    else:  
        fill_values[col] = df[col].mean()

df.fillna(fill_values, inplace=True)
```

```python
df
```

```python
numerical_df=df.select_dtypes(include=['int64','float64'])
corr=numerical_df.corr()[['Popularity']]
corr
```

```python
for col in numerical_df:
    plt.figure(figsize=(12,8))
    sns.scatterplot(x=col,y=numerical_df['Popularity'],data=df)
    plt.xlabel(col)
    plt.ylabel('Popularity')
    plt.title(f'Scatter plot of {col} vs Popularity')
    plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Feature Correlation Matrix")
plt.show()
correlated_features = corr['Popularity'].abs().sort_values(ascending=False)
print(correlated_features)
```

```python
from sklearn.preprocessing import StandardScaler
selected_features=['Instrumentalness','Mode','Tempo','Acousticness','Energy','Duration (ms)','Speechiness','Danceability']
scaler=StandardScaler()
df_scaled=scaler.fit_transform(df[selected_features])
df_scaled = pd.DataFrame(df_scaled, columns=selected_features)
```

```python
df_scaled
```

```python
from sklearn.model_selection import train_test_split
x=df_scaled
y=df['Popularity']
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f'Mean Squared Error: {mse}')
print(f'R² Score: {r2}')
```

```python
coefficients = pd.DataFrame(model.coef_, index=selected_features, columns=['Coefficient'])
coefficients = coefficients.sort_values(by='Coefficient', ascending=False)
print(coefficients)
```
